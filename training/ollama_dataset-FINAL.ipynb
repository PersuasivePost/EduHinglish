```python
# 7. Verify enrichment progress and sample outputs
import json

with open('data/ss9 (2).json', 'r', encoding='utf-8') as f:
    data = json.load(f)

filled = sum(1 for e in data if e.get('question_hinglish') and e.get('answer_hinglish'))
print(f'Total records: {len(data)} | Enriched records: {filled}')

print('\n--- Sample Enriched Records ---')
for i, e in enumerate(data[:3], 1):
    print(f'\n[Sample {i}] Chapter: {e.get("chapter")}')
    print('Q (English)  :', e.get('question_english'))
    print('Q (Hinglish) :', e.get('question_hinglish'))
    print('A (English)  :', str(e.get('answer_english'))[:120], '...')
    print('A (Hinglish) :', str(e.get('answer_hinglish'))[:120], '...')
```
